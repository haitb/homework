```python
# Bài 9: Nhập vào thời gian 1 công việc nào đó là x giây. 
# Hãy chuyển đổi và viết ra màn hình số thời gian trên dưới dạng bằng bao nhiêu giờ, bao nhiêu phút, bao nhiêu giây
x=int(input())
gio=x//3600
phut=(x-gio*3600)//60
giay=x-gio*3600-phut*60
print(f"{gio} giờ, {phut} phút, {giay} giây")
```

```python
# Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vào một điểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?
a,b,r,x,y=map(float, input().split())
import math
khoang_cach=math.sqrt((a-x)**2+(b-y)**2)
if khoang_cach-r==0:
    print("thuộc đường tròn")
elif khoang_cach-r>0:
    print("ngoài đường tròn")
else:
    print("trong đường tròn")
    
```
